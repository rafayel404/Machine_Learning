```python
import pandas as pd
import numpy as np
import time
from sklearn.linear_model import LinearRegression, SGDRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
from sklearn.datasets import make_regression

# Generate the data
X,y = make_regression(
    n_samples=100000,
    n_features=100,
    noise=10,
    random_state=42)

# Create a DataFrame to view it easily
df = pd.DataFrame(X, columns=[f'Feature_{i+1}' for i in range(100)])
df['Output'] = y

df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,:-1],df.iloc[:,-1],test_size=0.2,random_state=42)
```

```python
X_train.shape
```

```python
lr=LinearRegression()

ols_time_start = time.time()
lr.fit(X_train,y_train)
ols_time_end=time.time()

```

```python
sgd_time_start = time.time()
sgd = SGDRegressor()
sgd.fit(X_train,y_train)
sgd_time_end=time.time()
```

```python
print('Time taken by OLS is : ',ols_time_end-ols_time_start)
print('Time taken by Gradient Descent is : ',sgd_time_end-sgd_time_start)
```
